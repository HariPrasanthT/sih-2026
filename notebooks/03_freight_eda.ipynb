# Notebook 03: Freight Rate Exploratory Data Analysis
Investigates route-level rate distributions, corridor distances, and market indicator correlations.


```python
import pandas as pd
import matplotlib.pyplot as plt

df_master = pd.read_parquet('data/processed/master/training_master.parquet')
print(f'Training Master Dataset: {len(df_master):,} rows, {len(df_master.columns)} columns')
print('Corridor Groups:')
print(df_master.groupby(['origin_port', 'destination_port', 'vessel_class'])['freight_usd_per_ton'].agg(['count', 'mean', 'std', 'min', 'max']))
```

```python
# Boxplot of Freight by Vessel Class
plt.figure(figsize=(8, 5))
df_master.boxplot(column='freight_usd_per_ton', by='vessel_class')
plt.title('Freight Rate Distribution by Vessel Class')
plt.suptitle('')
plt.ylabel('USD / ton')
plt.show()
```
