# Notebook 01: Raw Maritime Data & Schema Audit
**Project:** SIH 26006 - Indian East Coast Freight Forecasting & Vessel Charter Recommendation System
Audits raw directory files, verifies column structures, checks null values, and inspects metadata manifest.


```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

schema_report_path = 'data/metadata/schema_report.json'
if os.path.exists(schema_report_path):
    with open(schema_report_path, 'r') as f:
        schema_data = json.load(f)
    print(f'Discovered Files Count: {schema_data.get("discovered_files_count")}')
    print(f'Summary: {json.dumps(schema_data.get("summary"), indent=2)}')
else:
    print('schema_report.json not found. Run scripts/01_discover_data.py first.')
```

```python
manifest_path = 'data/metadata/data_manifest.json'
if os.path.exists(manifest_path):
    with open(manifest_path, 'r') as f:
        manifest = json.load(f)
    df_man = pd.DataFrame(manifest.get('datasets', []))
    print(f'Registered datasets in manifest: {len(df_man)}')
    display(df_man[['dataset_name', 'scope', 'source_type', 'rows']]) if 'display' in globals() else print(df_man[['dataset_name', 'scope', 'source_type', 'rows']])
```
