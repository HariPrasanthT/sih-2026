# Notebook 02: Indian East Coast Port Physical Constraints & Congestion
Analyzes canonical discharge ports (Paradip, Dhamra, Visakhapatnam, etc.), physical limits, and derived congestion indices.


```python
import pandas as pd
import matplotlib.pyplot as plt

df_ports = pd.read_parquet('data/processed/ports/port_master.parquet')
df_constraints = pd.read_parquet('data/processed/ports/port_constraints.parquet')
print('Canonical Indian East Coast Ports:')
display(df_constraints) if 'display' in globals() else print(df_constraints)
```

```python
# Plot Maximum Draft and LOA Constraints
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
df_constraints.plot(x='canonical_name', y='max_draft_m', kind='bar', ax=ax1, color='#1f77b4', legend=False)
ax1.set_title('Max Permissible Draft by Port (m)')
ax1.set_ylabel('Draft (m)')
plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')

df_constraints.plot(x='canonical_name', y='max_loa_m', kind='bar', ax=ax2, color='#ff7f0e', legend=False)
ax2.set_title('Max Permissible LOA by Port (m)')
ax2.set_ylabel('LOA (m)')
plt.setp(ax2.get_xticklabels(), rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
