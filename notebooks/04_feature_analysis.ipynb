# Notebook 04: Feature Pipeline & Correlation Analysis
Examines freight lags, rolling momentum, correlation matrices, and verifies zero temporal leakage.


```python
import pandas as pd
import matplotlib.pyplot as plt

df_master = pd.read_parquet('data/processed/master/training_master.parquet')
feature_cols = [c for c in df_master.columns if 'freight' in c or 'bdi' in c or 'bunker' in c or 'congestion' in c]
corr = df_master[feature_cols].corr()
plt.figure(figsize=(10, 8))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(feature_cols)), feature_cols, rotation=90, fontsize=8)
plt.yticks(range(len(feature_cols)), feature_cols, fontsize=8)
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```
